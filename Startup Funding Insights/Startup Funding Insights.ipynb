```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"https://raw.githubusercontent.com/jerrengrimes/Projects/refs/heads/main/Startup%20Funding%20Insights/caplight_compiled.csv")
df["Total Funding2"] = df["Total Funding"].str.replace("$", "", regex=False)
df["Total Funding2"] = df["Total Funding2"].str.replace(",", "", regex=False)
df["Total Funding2"] = df["Total Funding2"].str.replace("-", "0", regex=False)
df["Total Funding2"] = df["Total Funding2"].str.replace("M", "*1000000", regex=False)
df["Total Funding2"] = df["Total Funding2"].str.replace("B", "*1000000000", regex=False)


def evaluate_math_string(val):

    if pd.isnull(val) or str(val).strip() in ['', '-', '--']:
        return None
    
    try:
        return eval(str(val))
    except (SyntaxError, NameError, TypeError):

        return None

df['Total Funding2'] = df['Total Funding2'].apply(evaluate_math_string)
sectors = df['Sector'].unique()
count = []
funding = []

for i in sectors:
    df2 = df[df['Sector'] == i]
    m = len(df2)
    count.append(m)
    s = sum(df2["Total Funding2"])
    funding.append(s)
    
df3 = pd.DataFrame({'sectors':sectors,'count':count,'funding':funding})
df3 = df3.sort_values(by='count', ascending = False)

%matplotlib tk
```

```python
plt.figure(figsize=(15, 8)) 
bars = plt.bar(df3['sectors'], df3['count'], color='red', edgecolor='k', width=0.6)

plt.xlabel('Company Type')
plt.ylabel('Number of Companies')
plt.xticks(rotation='vertical')
plt.bar_label(bars, padding=3)
plt.show()
plt.tight_layout()
```

```python
df4 = df3.sort_values(by='funding', ascending = False)

plt.figure(figsize=(15, 8)) 
bars = plt.bar(df4['sectors'], round(df4['funding']/1000000000,2), color='blue', edgecolor='k', width=0.6)

plt.xlabel('Company Type')
plt.ylabel('Funding Amount in $B')
plt.xticks(rotation='vertical')
plt.bar_label(bars, padding=3)
plt.show()
plt.tight_layout()
```
